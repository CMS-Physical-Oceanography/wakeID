# BoatFinder

Self-contained Colab workflow: build video URLs, download one video at a time, run the boat crossing detector, print crossing events, delete temporary video/images, and continue.

# READ ME

BoatFinder takes a specified start and end date of interest and analyzes all available CMS south-view camera videos within those dates. The camera time is calculated from the video filename plus the crossing frame offset; the filename timestamp is slightly behind the true displayed camera time, so a manual time correction is applied with `camera_time_offset_seconds`. In the current state of the code, the corrected `time` should be accurate to approximately +/- 1 second. A region of interest (ROI) is created around the center portion of the frame, and YOLO is used to detect boats inside that ROI. The code records an event when a tracked boat crosses the selected vertical x-line. A GCP survey is used to create a homography matrix that converts full-frame pixel coordinates into NC State Plane feet/UTM-style Easting and Northing coordinates; tidal influence has not been adjusted for. For each crossing, the frame before the boat crosses the x-line and the frame after it crosses the x-line are captured. The bottom-center point of each YOLO box is converted into real-world coordinates, and the Pythagorean theorem is used to estimate the distance traveled between those two detections. The elapsed frame count is divided by the video FPS to calculate elapsed time, and speed is calculated as distance traveled over elapsed time. A line is also created between the before and after boat positions. Where that line intersects the x-line, the pixel coordinate is saved, converted into real-world coordinates, and compared to the pressure sensor location with the Pythagorean theorem to estimate boat distance from the sensor.

For faster processing, the kernel should be changed to GCU computing. 

Checkpointing: the final processing loop writes results to the CSV after each full day is processed. This means completed days are already saved in `results_csv_path` before the next day begins, so a crash or disconnect should only risk the current unfinished day. The loop prints a checkpoint message after each day showing how many rows were appended and how many videos were skipped.

## Main Variables (can be modified in "Configuration, Model, and Calibration" block):

# Modify these:

- `start_date`: first archive date to process, formatted as `YYYY-MM-DD`.
- `end_date`: final archive date to process, formatted as `YYYY-MM-DD`.
- `results_csv_path`: Google Drive path where the final CSV is written.
- `mat_path`: path to `data.mat`, which stores the GCP image and real-world coordinate pairs.
- `pressure_sensor_east_ft`: pressure sensor Easting coordinate in NC State Plane feet.
- `pressure_sensor_north_ft`: pressure sensor Northing coordinate in NC State Plane feet.

# Should not need modification:

- `VIDEO_FPS`: fixed video frame rate used to convert frame counts into seconds. Should consistently be 5. 
- `search_seconds`: how often search mode checks for boats.
- `rewind_seconds`: how far back track mode rewinds after search mode finds a boat.
- `track_stride_frames`: frame spacing used while tracking an active boat event.
- `no_boat_timeout_seconds`: time with no boat detections before returning to search mode.
- `duplicate_crossing_cooldown_seconds`: suppresses extra crossing events detected within this many seconds of the last accepted crossing.
- `camera_time_offset_seconds`: manual timestamp correction added after filename/frame timing; negative values move the output time earlier.
- `boat_class_id`: YOLO class ID used for boats.
- `x1, y1, x2, y2`: ROI coordinates in the original full-frame image.
- `max_match_distance`: maximum pixel distance allowed when matching detections between frames.
- `max_missing_frames`: number of missed track updates allowed before removing a custom track.
- `direction_penalty`: extra matching cost applied when a detection moves opposite the expected direction.
- `duplicate_x_center_threshold`: removes duplicate detections with nearly identical x-centers.
- `minimum_track_box_area`: minimum YOLO box area accepted during track mode.
- `maximum_valid_boat_length_ft`: skips a crossing when both before/after boat-length estimates are greater than this value.
- `slow_boat_speed_threshold_ft_per_s`: speed threshold used to identify slow-boat duplicate crossing risk.
- `slow_boat_cooldown_seconds`: cooldown time used to suppress repeated slow-boat crossings.
- `search_confidence`: YOLO confidence threshold used during search mode.
- `track_confidence`: YOLO confidence threshold used during track mode.
- `search_model`: YOLO model used for sparse search.
- `track_model`: YOLO model used for active tracking.
- `pixel_to_utm_matrix`: homography matrix that converts image pixels to real-world coordinates.

## CSV Outputs

- `video_url`: source video URL processed for this row. Every URL within the date range is pasted, even if there is no captured boat.
- `skip_reason`: reason a video was skipped; blank for successful crossing events.
- `time`: crossing time calculated from the video filename timestamp minus 4 hours, plus `frame_index / VIDEO_FPS`, plus `camera_time_offset_seconds`.
- `frame_index`: video frame index where the crossing was detected.
- `direction`: crossing direction, either `left_to_right` or `right_to_left`.
- `previous_bottom_center_pixel`: full-frame pixel coordinate of the boat's bottom-center point before crossing.
- `current_bottom_center_pixel`: full-frame pixel coordinate of the boat's bottom-center point after crossing.
- `crossing_intersection_pixel`: full-frame pixel coordinate where the boat path line intersects the x-line.
- `crossing_step_elapsed_seconds`: elapsed time between the previous and current boat detections used for the crossing.
- `average_box_width_pixels`: average YOLO box width from the before and after crossing detections.
- `average_box_height_pixels`: average YOLO box height from the before and after crossing detections.
- `previous_box_boat_length_ft`: estimated real-world length of the previous YOLO box bottom edge (UTM conversion/pythagorean theorem).
- `current_box_boat_length_ft`: estimated real-world length of the current YOLO box bottom edge (UTM conversion/pythagorean theorem).
- `average_boat_length_ft`: average of the previous and current box-based boat length estimates (UTM conversion/pythagorean theorem).
- `crossing_intersection_east`: real-world Easting coordinate of the path/x-line intersection (UTM conversion).
- `crossing_intersection_north`: real-world Northing coordinate of the path/x-line intersection (UTM conversion).
- `distance_to_pressure_sensor_ft`: distance from the path/x-line intersection to the pressure sensor, in feet (UTM conversion/pythagorean theorem).
- `crossing_step_distance_ft`: real-world distance traveled between the previous and current bottom-center detections, in feet (UTM conversion/pythagorean theorem).
- `crossing_step_speed_ft_per_s`: estimated speed between the previous and current detections, in feet per second.

## URL And Download Helpers

This cell defines helper functions for building the list of archive video URLs, downloading one video at a time, and drawing extended path lines inside the ROI.

```python
# Base archive folder for the CMS dock south camera. The date folders are added later.
DEFAULT_BASE_URL = "https://stage-ams.srv.axds.co/archive/mp4/uncw/cms_dock_south/"


def _parse_date(value):
    # Accept either a Python date object or a string formatted like "2022-10-01".
    if isinstance(value, date):
        return value

    return datetime.strptime(value, "%Y-%m-%d").date()


def _date_range(start_date, end_date):
    # Generate every date from start_date through end_date, including both endpoints.
    current_date = _parse_date(start_date)
    final_date = _parse_date(end_date)

    if current_date > final_date:
        raise ValueError("start_date must be before or equal to end_date")

    while current_date <= final_date:
        yield current_date
        current_date += timedelta(days=1)


def get_video_urls(start_date, end_date, base_url=DEFAULT_BASE_URL):
    # Visit each daily archive page and collect every MP4 link found on that page.
    base_url = base_url.rstrip("/") + "/"
    video_urls = []

    for current_date in _date_range(start_date, end_date):
        # Daily pages follow the pattern base_url/YYYY/MM/DD/.
        day_url = urljoin(
            base_url,
            f"{current_date:%Y}/{current_date:%m}/{current_date:%d}/",
        )

        # Download the small HTML directory listing for this day.
        response = requests.get(day_url, timeout=30)
        response.raise_for_status()

        # Pull out MP4 filenames and keep only the first copy of each file.
        # The archive HTML can show each filename twice: once in href and once as link text.
        filenames = list(dict.fromkeys(re.findall(r"cms.*?\.mp4", response.text)))

        for filename in filenames:
            video_urls.append(urljoin(day_url, filename))

    return video_urls


def download_video(video_url, output_folder="/content"):
    # Download one MP4 into temporary Colab storage and return its local path.
    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)

    filename = video_url.rstrip("/").split("/")[-1]
    local_path = output_folder / filename

    # Stream the video in chunks so the full MP4 is not loaded into memory.
    with requests.get(video_url, stream=True, timeout=30) as response:
        response.raise_for_status()

        with open(local_path, "wb") as file:
            for chunk in response.iter_content(chunk_size=1024 * 1024):
                if chunk:
                    file.write(chunk)

    return str(local_path)


def get_video_start_time(video_source):
    # Pull the timestamp out of names like cms_dock_south-2022-10-01-110405Z.mp4.
    filename = os.path.basename(video_source.rstrip("/"))
    match = re.search(r"(\d{4})-(\d{2})-(\d{2})-(\d{2})(\d{2})(\d{2})Z", filename)

    # Stop clearly if the video filename does not contain the expected timestamp pattern.
    if match is None:
        raise ValueError(f"Could not find video timestamp in filename: {filename}")

    # Convert the filename timestamp into a datetime object.
    video_start_time = datetime(
        int(match.group(1)),
        int(match.group(2)),
        int(match.group(3)),
        int(match.group(4)),
        int(match.group(5)),
        int(match.group(6)),
    )

    # The filename timestamp ends in Z; subtract 4 hours for the desired EST-style output.
    return video_start_time - timedelta(hours=4)


def get_camera_time_from_frame(video_start_time, frame_index, fps, time_offset_seconds=0):
    # Convert this frame number into seconds after the video started.
    seconds_after_start = frame_index / fps

    # Add the frame offset and the manual correction to the filename-derived start time.
    camera_time = video_start_time + timedelta(
        seconds=seconds_after_start + time_offset_seconds
    )

    # Return a consistent timestamp string for crossing_events.
    return camera_time.strftime("%m/%d/%Y %H:%M:%S")


def line_intersections_with_rect(point1, point2, rect):
    # Extend a two-point path line until it hits the ROI rectangle boundary.
    x_min, y_min, x_max, y_max = rect
    x_a, y_a = point1
    x_b, y_b = point2
    dx = x_b - x_a
    dy = y_b - y_a
    intersections = []

    if dx != 0:
        for x_edge in (x_min, x_max):
            t = (x_edge - x_a) / dx
            y_edge = y_a + t * dy
            if y_min <= y_edge <= y_max:
                intersections.append((int(round(x_edge)), int(round(y_edge))))

    if dy != 0:
        for y_edge in (y_min, y_max):
            t = (y_edge - y_a) / dy
            x_edge = x_a + t * dx
            if x_min <= x_edge <= x_max:
                intersections.append((int(round(x_edge)), int(round(y_edge))))

    unique_intersections = []
    for point in intersections:
        if point not in unique_intersections:
            unique_intersections.append(point)

    if len(unique_intersections) >= 2:
        return unique_intersections[0], unique_intersections[1]

    return point1, point2
```

## Configuration, Model, And Calibration

This cell sets the date range and detector parameters, loads YOLO once, reads `data.mat` from Google Drive, and builds the pixel-to-real-world homography.

```python
# Modify these:
# Change these two dates to control which archive videos are processed.
start_date = "2022-10-11"
end_date = "2022-10-11"

# Final CSV output path in Google Drive.
results_csv_path = "/content/drive/MyDrive/Colabdata/boat_crossing_events.csv"

# Calibration file in Google Drive. This is loaded once before the video loop.
mat_path = "/content/drive/MyDrive/Colabdata/data.mat"

# Pressure sensor location in NC State Plane feet.
pressure_sensor_east_ft = 2344242.3
pressure_sensor_north_ft = 143745.445


# Should not need modification:
# Download only one temporary video at a time into Colab's local runtime disk.
temporary_video_folder = "/content"

# Crossing debug images are saved here during one video, then deleted after that video.
image_output_folder = "/content/Boatimages"

# VIDEO_FPS is intentionally fixed for now to match the original pythontester.py behavior.
VIDEO_FPS = 5

# Search/track timing settings.
search_seconds = 2
rewind_seconds = 3
track_stride_frames = 5
no_boat_timeout_seconds = 5

# Suppress any extra crossing event detected within this many seconds of the last accepted crossing.
duplicate_crossing_cooldown_seconds = 4

# Manual correction applied to each calculated camera timestamp.
# Negative values move the output time earlier.
camera_time_offset_seconds = -5

# YOLO's COCO class ID for boat is 8.
boat_class_id = 8

# ROI coordinates in the original full video frame.
x1, y1, x2, y2 = 315, 340, 750, 675

# Custom tracker settings.
max_match_distance = 250
max_missing_frames = 5
direction_penalty = 100
duplicate_x_center_threshold = 2.5
minimum_track_box_area = 160

# If both before/after boat-length estimates exceed this value, skip the crossing.
maximum_valid_boat_length_ft = 100

# Slow boats can produce multiple YOLO boxes on the same hull.
# After accepting a slow crossing, suppress additional crossings briefly.
slow_boat_speed_threshold_ft_per_s = 15
slow_boat_cooldown_seconds = 12

# YOLO confidence thresholds and model choices.
# Search uses nano for speed; track uses medium for better active-event sensitivity.
search_confidence = 0.10
track_confidence = 0.20
search_model = YOLO("yolov8n.pt")
track_model = YOLO("yolov8m.pt")

# Load calibration once. Reuse this same homography for every video.
# The GCP points convert image pixels into real-world East/North coordinates.
mat_data = scipy.io.loadmat(mat_path, squeeze_me=True, struct_as_record=False)
both_coord = mat_data["bothCoord"]

pixel_points = []
utm_points = []

# data.mat stores gcp1 through gcp22 in the bothCoord MATLAB structure.
for gcp_number in range(1, 23):
    gcp = getattr(both_coord, f"gcp{gcp_number}")
    pixel_points.append([float(gcp.Ximg), float(gcp.Yimg)])
    utm_points.append([float(gcp.East), float(gcp.North)])

pixel_points = np.array(pixel_points, dtype=np.float32)
utm_points = np.array(utm_points, dtype=np.float32)

# Estimate the projective transform from image pixels to UTM/state-plane coordinates.
pixel_to_utm_matrix, homography_inlier_mask = cv2.findHomography(
    pixel_points,
    utm_points,
    cv2.RANSAC,
)

if pixel_to_utm_matrix is None:
    raise RuntimeError("Could not compute pixel-to-UTM homography from GCP points.")


def project_pixel_to_utm(pixel_point):
    # Convert one full-frame pixel coordinate into projected real-world coordinates.
    pixel_array = np.array([[pixel_point]], dtype=np.float32)
    utm_array = cv2.perspectiveTransform(pixel_array, pixel_to_utm_matrix)
    east = float(utm_array[0][0][0])
    north = float(utm_array[0][0][1])
    return east, north
```

## Process One Video

This cell defines `process_video(video_path)`, which contains the moved `pythontester.py` logic for one downloaded video and returns that video's `crossing_events`.

```python
def process_video(video_path, video_url=None):
    # Process exactly one downloaded video and return its crossing event list.
    start_time = time.perf_counter()

    # OpenCV reads from the local file downloaded by download_video().
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise RuntimeError(f"Could not open video: {video_path}")

    # Create the temporary image folder for annotated crossing snapshots.
    os.makedirs(image_output_folder, exist_ok=True)

    # Reset tracking state for every video so tracks/events do not leak between files.
    active_tracks = {}
    next_track_id = 1
    crossing_events = []
    crossing_count = 0
    mode = "search"
    track_start_frame = None
    last_boat_frame = None
    last_crossing_event_frame = None
    last_slow_boat_crossing_frame = None

    # Convert time-based settings into frame counts for this video.
    video_fps = VIDEO_FPS

    # Get the corrected video start time from the filename instead of using OCR.
    video_start_time = get_video_start_time(video_url or video_path)

    search_interval_frames = video_fps * search_seconds
    rewind_frames = video_fps * rewind_seconds
    no_boat_timeout_frames = video_fps * no_boat_timeout_seconds
    duplicate_crossing_cooldown_frames = video_fps * duplicate_crossing_cooldown_seconds
    slow_boat_cooldown_frames = video_fps * slow_boat_cooldown_seconds

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame_index = int(cap.get(cv2.CAP_PROP_POS_FRAMES)) - 1
        seconds = frame_index / video_fps

        if mode == "search":
            # Search mode runs YOLO sparsely until a boat appears in the ROI.
            if frame_index % search_interval_frames != 0:
                continue

            # Only send the ROI to YOLO to reduce computation.
            roi = frame[y1:y2, x1:x2]

            search_results = search_model.predict(
                roi,
                classes=[boat_class_id],
                conf=search_confidence,
                verbose=False,
            )

            boat_detected = any(len(result.boxes) > 0 for result in search_results)

            if not boat_detected:
                continue

            # Rewind a few seconds so track mode can capture the lead-up to the crossing.
            track_start_frame = max(0, frame_index - rewind_frames)
            last_boat_frame = frame_index
            active_tracks.clear()
            next_track_id = 1

            cap.set(cv2.CAP_PROP_POS_FRAMES, track_start_frame)
            mode = "track"
            continue

        if mode == "track":
            # Track mode runs YOLO more often and checks whether tracked boats cross the line.
            # If the scene has been empty long enough, go back to sparse search mode.
            if frame_index - last_boat_frame >= no_boat_timeout_frames:
                print(
                    f"leaving track mode - frame {frame_index}, seconds {seconds:.2f}, "
                    f"last_boat_frame {last_boat_frame}"
                )
                mode = "search"
                track_start_frame = None
                last_boat_frame = None
                active_tracks.clear()
                next_track_id = 1
                continue

            # Skip non-stride frames to reduce YOLO calls while tracking.
            if (frame_index - track_start_frame) % track_stride_frames != 0:
                continue

            print(f"track yolo - frame {frame_index}, seconds {seconds:.2f}")

            roi = frame[y1:y2, x1:x2]
            roi_height, roi_width = roi.shape[:2]
            line_x = roi_width / 2

            track_results = track_model.predict(
                roi,
                classes=[boat_class_id],
                conf=track_confidence,
                verbose=False,
            )

            total_boxes = sum(len(result.boxes) for result in track_results)
            print(f"track detections - frame {frame_index}, boxes {total_boxes}")

            # Convert YOLO boxes into center points and ROI-relative boxes for custom tracking.
            current_detections = []
            for result in track_results:
                for box in result.boxes:
                    x1b, y1b, x2b, y2b = box.xyxy[0]

                    box_width = float(x2b - x1b)
                    box_height = float(y2b - y1b)
                    box_area = box_width * box_height
                    if box_area < minimum_track_box_area:
                        continue

                    x_center = float((x1b + x2b) / 2)
                    y_center = float((y1b + y2b) / 2)
                    current_detections.append({
                        "x": x_center,
                        "y": y_center,
                        "box": (
                            int(x1b),
                            int(y1b),
                            int(x2b),
                            int(y2b),
                        ),
                    })

                    print(
                        f"track box - frame {frame_index}, "
                        f"x_center {x_center:.1f}, line_x {line_x:.1f}"
                    )

            # Remove near-duplicate detections so one boat is less likely to become two tracks.
            filtered_detections = []
            for detection in current_detections:
                duplicate_detection = any(
                    abs(detection["x"] - kept_detection["x"]) <= duplicate_x_center_threshold
                    for kept_detection in filtered_detections
                )

                if not duplicate_detection:
                    filtered_detections.append(detection)

            current_detections = filtered_detections

            unmatched_detection_indices = set(range(len(current_detections)))
            unmatched_track_ids = set(active_tracks.keys())
            matches = []

            # Match current detections to existing tracks using predicted position and direction.
            for detection_index, detection in enumerate(current_detections):
                best_track_id = None
                best_score = None

                for track_id in unmatched_track_ids:
                    track = active_tracks[track_id]
                    predicted_x = track["x"] + track["velocity_x"]
                    predicted_y = track["y"] + track["velocity_y"]
                    dx = detection["x"] - predicted_x
                    dy = detection["y"] - predicted_y
                    match_distance = (dx ** 2 + dy ** 2) ** 0.5
                    score = match_distance

                    movement_x = detection["x"] - track["x"]
                    if track["velocity_x"] > 0 and movement_x < 0:
                        score += direction_penalty
                    elif track["velocity_x"] < 0 and movement_x > 0:
                        score += direction_penalty

                    if best_score is None or score < best_score:
                        best_score = score
                        best_track_id = track_id

                if best_track_id is not None and best_score <= max_match_distance:
                    matches.append((best_track_id, detection_index))
                    unmatched_track_ids.remove(best_track_id)
                    unmatched_detection_indices.remove(detection_index)

            # For matched tracks, compare previous/current x position against the crossing line.
            for track_id, detection_index in matches:
                track = active_tracks[track_id]
                detection = current_detections[detection_index]
                prev_x = track["x"]
                current_x = detection["x"]
                crossed_lr = prev_x < line_x and current_x >= line_x
                crossed_rl = prev_x > line_x and current_x <= line_x

                print(
                    f"crossing check - frame {frame_index}, track_id {track_id}, "
                    f"prev_x {prev_x:.1f}, current_x {current_x:.1f}, "
                    f"velocity_x {track['velocity_x']:.1f}, "
                    f"crossed_lr {crossed_lr}, crossed_rl {crossed_rl}"
                )

                if crossed_lr or crossed_rl:
                    # Skip duplicate crossing events that happen too close to the last accepted crossing.
                    in_duplicate_crossing_cooldown = (
                        last_crossing_event_frame is not None
                        and frame_index - last_crossing_event_frame <= duplicate_crossing_cooldown_frames
                    )

                    if in_duplicate_crossing_cooldown:
                        continue

                    # A crossing event was detected; calculate geometry and save a debug image.
                    direction = "left_to_right" if crossed_lr else "right_to_left"

                    annotated_frame = frame.copy()

                    prev_box = track["box"]
                    current_box = detection["box"]
                    prev_box_full = (
                        x1 + prev_box[0],
                        y1 + prev_box[1],
                        x1 + prev_box[2],
                        y1 + prev_box[3],
                    )
                    current_box_full = (
                        x1 + current_box[0],
                        y1 + current_box[1],
                        x1 + current_box[2],
                        y1 + current_box[3],
                    )

                    prev_box_width = prev_box_full[2] - prev_box_full[0]
                    prev_box_height = prev_box_full[3] - prev_box_full[1]
                    current_box_width = current_box_full[2] - current_box_full[0]
                    current_box_height = current_box_full[3] - current_box_full[1]
                    average_box_width_pixels = (prev_box_width + current_box_width) / 2
                    average_box_height_pixels = (prev_box_height + current_box_height) / 2

                    prev_bottom_left = (prev_box_full[0], prev_box_full[3])
                    prev_bottom_right = (prev_box_full[2], prev_box_full[3])
                    current_bottom_left = (current_box_full[0], current_box_full[3])
                    current_bottom_right = (current_box_full[2], current_box_full[3])

                    prev_left_east, prev_left_north = project_pixel_to_utm(prev_bottom_left)
                    prev_right_east, prev_right_north = project_pixel_to_utm(prev_bottom_right)
                    current_left_east, current_left_north = project_pixel_to_utm(current_bottom_left)
                    current_right_east, current_right_north = project_pixel_to_utm(current_bottom_right)

                    previous_box_boat_length_ft = (
                        (prev_right_east - prev_left_east) ** 2
                        + (prev_right_north - prev_left_north) ** 2
                    ) ** 0.5

                    current_box_boat_length_ft = (
                        (current_right_east - current_left_east) ** 2
                        + (current_right_north - current_left_north) ** 2
                    ) ** 0.5

                    average_boat_length_ft = (
                        previous_box_boat_length_ft + current_box_boat_length_ft
                    ) / 2

                    # Skip likely bad detections when both length estimates are unrealistically large.
                    both_boat_lengths_too_large = (
                        previous_box_boat_length_ft > maximum_valid_boat_length_ft
                        and current_box_boat_length_ft > maximum_valid_boat_length_ft
                    )

                    if both_boat_lengths_too_large:
                        continue

                    prev_bottom_center = (
                        int((prev_box_full[0] + prev_box_full[2]) / 2),
                        prev_box_full[3],
                    )
                    current_bottom_center = (
                        int((current_box_full[0] + current_box_full[2]) / 2),
                        current_box_full[3],
                    )

                    full_line_x = x1 + line_x
                    path_dx = current_bottom_center[0] - prev_bottom_center[0]
                    path_dy = current_bottom_center[1] - prev_bottom_center[1]

                    if path_dx != 0:
                        intersection_t = (full_line_x - prev_bottom_center[0]) / path_dx
                        crossing_intersection_y = prev_bottom_center[1] + intersection_t * path_dy
                    else:
                        crossing_intersection_y = current_bottom_center[1]

                    crossing_intersection_pixel = (
                        int(round(full_line_x)),
                        int(round(crossing_intersection_y)),
                    )

                    crossing_intersection_east, crossing_intersection_north = project_pixel_to_utm(
                        crossing_intersection_pixel
                    )

                    distance_to_pressure_sensor_ft = (
                        (crossing_intersection_east - pressure_sensor_east_ft) ** 2
                        + (crossing_intersection_north - pressure_sensor_north_ft) ** 2
                    ) ** 0.5

                    prev_east, prev_north = project_pixel_to_utm(prev_bottom_center)
                    current_east, current_north = project_pixel_to_utm(current_bottom_center)

                    crossing_step_distance_ft = (
                        (current_east - prev_east) ** 2
                        + (current_north - prev_north) ** 2
                    ) ** 0.5

                    crossing_step_frame_delta = frame_index - track["last_seen_frame"]
                    crossing_step_seconds = crossing_step_frame_delta / video_fps

                    if crossing_step_seconds > 0:
                        crossing_step_speed_ft_per_s = crossing_step_distance_ft / crossing_step_seconds
                    else:
                        crossing_step_speed_ft_per_s = 0

                    is_slow_boat_crossing = (
                        crossing_step_speed_ft_per_s < slow_boat_speed_threshold_ft_per_s
                    )
                    in_slow_boat_cooldown = (
                        last_slow_boat_crossing_frame is not None
                        and frame_index - last_slow_boat_crossing_frame <= slow_boat_cooldown_frames
                    )

                    if is_slow_boat_crossing and in_slow_boat_cooldown:
                        continue

                    if is_slow_boat_crossing:
                        last_slow_boat_crossing_frame = frame_index

                    # This event passed all duplicate filters, so count it and reserve its image name.
                    crossing_count += 1
                    last_crossing_event_frame = frame_index
                    image_path = os.path.join(image_output_folder, f"crossing_{crossing_count}.jpg")

                    cv2.rectangle(
                        annotated_frame,
                        (prev_box_full[0], prev_box_full[1]),
                        (prev_box_full[2], prev_box_full[3]),
                        (0, 0, 255),
                        3,
                    )
                    cv2.rectangle(
                        annotated_frame,
                        (current_box_full[0], current_box_full[1]),
                        (current_box_full[2], current_box_full[3]),
                        (0, 255, 0),
                        3,
                    )

                    line_start, line_end = line_intersections_with_rect(
                        prev_bottom_center,
                        current_bottom_center,
                        (x1, y1, x2, y2),
                    )
                    cv2.line(
                        annotated_frame,
                        line_start,
                        line_end,
                        (0, 0, 0),
                        3,
                    )

                    # Use the filename start time plus this frame's offset to avoid unreliable OCR.
                    crossing_time = get_camera_time_from_frame(
                        video_start_time,
                        frame_index,
                        video_fps,
                        camera_time_offset_seconds,
                    )

                    # Store the annotated frame locally and record the measured event values.
                    cv2.imwrite(image_path, annotated_frame)
                    crossing_events.append({
                        "video_url": video_url or video_path,
                        "frame_index": frame_index,
                        "direction": direction,
                        "time": crossing_time,
                        "previous_bottom_center_pixel": prev_bottom_center,
                        "current_bottom_center_pixel": current_bottom_center,
                        "crossing_intersection_pixel": crossing_intersection_pixel,
                        "crossing_step_elapsed_seconds": crossing_step_seconds,
                        "average_box_width_pixels": average_box_width_pixels,
                        "average_box_height_pixels": average_box_height_pixels,
                        "previous_box_boat_length_ft": previous_box_boat_length_ft,
                        "current_box_boat_length_ft": current_box_boat_length_ft,
                        "average_boat_length_ft": average_boat_length_ft,
                        "crossing_intersection_east": crossing_intersection_east,
                        "crossing_intersection_north": crossing_intersection_north,
                        "distance_to_pressure_sensor_ft": distance_to_pressure_sensor_ft,
                        "crossing_step_distance_ft": crossing_step_distance_ft,
                        "crossing_step_speed_ft_per_s": crossing_step_speed_ft_per_s,
                    })

                active_tracks[track_id] = {
                    "x": detection["x"],
                    "y": detection["y"],
                    "box": detection["box"],
                    "velocity_x": detection["x"] - track["x"],
                    "velocity_y": detection["y"] - track["y"],
                    "last_seen_frame": frame_index,
                    "missing_count": 0,
                }

            # Any unmatched detection starts a new custom track.
            for detection_index in unmatched_detection_indices:
                detection = current_detections[detection_index]
                active_tracks[next_track_id] = {
                    "x": detection["x"],
                    "y": detection["y"],
                    "box": detection["box"],
                    "velocity_x": 0,
                    "velocity_y": 0,
                    "last_seen_frame": frame_index,
                    "missing_count": 0,
                }
                print(f"new custom track - frame {frame_index}, track_id {next_track_id}")
                next_track_id += 1

            # Unmatched tracks age out after too many missed detections.
            for track_id in list(unmatched_track_ids):
                active_tracks[track_id]["missing_count"] += 1
                print(
                    f"missing custom track - frame {frame_index}, track_id {track_id}, "
                    f"missing_count {active_tracks[track_id]['missing_count']}"
                )

                if active_tracks[track_id]["missing_count"] > max_missing_frames:
                    print(f"removing custom track - frame {frame_index}, track_id {track_id}")
                    del active_tracks[track_id]

            if current_detections:
                last_boat_frame = frame_index

    # Release the video file before the outer loop deletes it.
    cap.release()

    elapsed_time = time.perf_counter() - start_time
    print(f"crossing_count: {crossing_count}")
    print(f"Elapsed time: {elapsed_time:.2f} seconds")

    return crossing_events
```

## Download, Process, Checkpoint, And Clean Up

This cell processes one day at a time. After a full day finishes, it appends that day's rows to the CSV as a checkpoint, deletes temporary files, and then moves to the next day.


```python
# Main pipeline loop: process one day at a time and append a checkpoint after each day.
# The process_video() debug prints are captured so this cell stays quiet during long runs.
# Bad or unopenable video URLs are logged and skipped so one broken MP4 does not stop the batch.
import contextlib
import io

csv_column_order = [
    "video_url",
    "skip_reason",
    "time",
    "frame_index",
    "direction",
    "previous_bottom_center_pixel",
    "current_bottom_center_pixel",
    "crossing_intersection_pixel",
    "crossing_step_elapsed_seconds",
    "average_box_width_pixels",
    "average_box_height_pixels",
    "previous_box_boat_length_ft",
    "current_box_boat_length_ft",
    "average_boat_length_ft",
    "crossing_intersection_east",
    "crossing_intersection_north",
    "distance_to_pressure_sensor_ft",
    "crossing_step_distance_ft",
    "crossing_step_speed_ft_per_s",
]

total_saved_rows = 0
skipped_videos = []

for current_day in _date_range(start_date, end_date):
    day_crossing_events = []
    day_skipped_videos = []
    day_label = current_day.strftime("%Y-%m-%d")

    try:
        video_urls_for_day = get_video_urls(current_day, current_day)
    except Exception as error:
        skipped_day = {
            "video_url": f"archive day {day_label}",
            "skip_reason": f"Could not collect video URLs: {error}",
        }
        day_crossing_events.append(skipped_day)
        day_skipped_videos.append(skipped_day)
        video_urls_for_day = []

    for video_url in video_urls_for_day:
        video_path = None

        try:
            # video_path is the changing input that replaces pythontester.py's old hardcoded path.
            video_path = download_video(video_url, output_folder=temporary_video_folder)

            # Skip downloads that produced no usable local file.
            if not os.path.exists(video_path) or os.path.getsize(video_path) == 0:
                raise RuntimeError(f"Downloaded video is missing or empty: {video_path}")

            # This is where the moved pythontester logic runs for the current video.
            with contextlib.redirect_stdout(io.StringIO()):
                crossing_events = process_video(video_path, video_url=video_url)

            # Keep one row for videos with no crossing so every processed URL appears in the CSV.
            if not crossing_events:
                crossing_events = [{"video_url": video_url}]

            day_crossing_events.extend(crossing_events)

        except Exception as error:
            # Record the failed URL and continue with the rest of the batch.
            skipped_video = {
                "video_url": video_url,
                "skip_reason": str(error),
            }
            day_skipped_videos.append(skipped_video)
            day_crossing_events.append(skipped_video)
            print(f"Skipping video: {video_url}")
            print(f"Reason: {error}")

        finally:
            # Always delete the large downloaded MP4, even if processing fails partway through.
            if video_path is not None and os.path.exists(video_path):
                os.remove(video_path)

            # Clear temporary crossing images before the next video starts.
            if os.path.exists(image_output_folder):
                shutil.rmtree(image_output_folder)

    # Append this day's rows to the CSV so completed days survive crashes/disconnects.
    if day_crossing_events:
        day_results_df = pd.DataFrame(day_crossing_events)
        day_results_df = day_results_df.reindex(columns=csv_column_order)

        write_header = not os.path.exists(results_csv_path)
        day_results_df.to_csv(
            results_csv_path,
            mode="a",
            header=write_header,
            index=False,
        )

        total_saved_rows += len(day_crossing_events)
        skipped_videos.extend(day_skipped_videos)

    print(
        f"Checkpoint saved for {day_label}: "
        f"{len(day_crossing_events)} rows appended, "
        f"{len(day_skipped_videos)} skipped videos."
    )

print(f"Finished processing. Total rows appended this run: {total_saved_rows}")
print(f"CSV path: {results_csv_path}")

if skipped_videos:
    print(f"Skipped {len(skipped_videos)} videos total:")
    for skipped_video in skipped_videos:
        print(f"- {skipped_video['video_url']} :: {skipped_video['skip_reason']}")
```
